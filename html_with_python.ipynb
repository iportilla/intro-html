## Part 2: Rendering HTML with Python
An HTML page is just text, so Python can create one like any other string.

### Example 1: Write and Display a Page
The `.py` version opens the page in your browser. Here, the notebook displays it below the cell.

````python
from pathlib import Path
from IPython.display import HTML, display

html = """<!DOCTYPE html>
<html>
<head>
    <title>Made with Python</title>
</head>
<body>
    <h1>This page was created by Python</h1>
    <p>Python wrote this HTML to a file.</p>
</body>
</html>
"""

page = Path("python_page.html")
page.write_text(html, encoding="utf-8")

display(HTML(page.read_text(encoding="utf-8")))
````

### Example 2: Build a Page from Data
A list of dictionaries stores the data, a `for` loop creates each table row, and f-strings insert the values.

````python
from pathlib import Path
from IPython.display import HTML, display

sales = [
    {"product": "Notebook", "units": 120, "price": 3.50},
    {"product": "Backpack", "units": 45, "price": 39.99},
    {"product": "Water Bottle", "units": 80, "price": 14.25},
]

rows = ""
total = 0

for item in sales:
    revenue = item["units"] * item["price"]
    total += revenue
    rows += f"""
    <tr>
        <td>{item["product"]}</td>
        <td>{item["units"]}</td>
        <td>${revenue:,.2f}</td>
    </tr>"""

html = f"""<!DOCTYPE html>
<html>
<head>
    <title>Sales Report</title>
</head>
<body>
    <h1>Sales Report</h1>
    <table border="1" cellpadding="8">
        <tr>
            <th>Product</th>
            <th>Units</th>
            <th>Revenue</th>
        </tr>{rows}
    </table>
    <p><strong>Total revenue: ${total:,.2f}</strong></p>
</body>
</html>
"""

page = Path("sales_report.html")
page.write_text(html, encoding="utf-8")

display(HTML(page.read_text(encoding="utf-8")))
````

## Part 3: Introduction to UI with Streamlit
Streamlit apps run as their own web page, not inside a notebook. Each cell below saves an app as a `.py` file. To run one, open a terminal in this folder and use:

```bash
python -m streamlit run app.py
```

Press `Ctrl+C` in the terminal to stop the app.

### Example: Tip Calculator

````python
%%writefile app.py
import streamlit as st

st.title("Tip Calculator")
st.write("Enter your bill to see the tip and total.")

name = st.text_input("Your name")
bill = st.number_input("Bill amount ($)", min_value=0.0, value=25.0, step=1.0)
tip_percent = st.slider("Tip (%)", min_value=0, max_value=30, value=20)

tip = bill * tip_percent / 100
total = bill + tip

st.write(f"Tip: ${tip:.2f}")
st.write(f"Total: ${total:.2f}")

if st.button("Say thanks"):
    if name:
        st.success(f"Thanks, {name}!")
    else:
        st.warning("Enter your name first.")
````

### Remembering Values with `st.session_state`
Streamlit reruns your script from top to bottom after every interaction, so regular variables start over. Values in `st.session_state` are kept. Run this one with `python -m streamlit run click_counter.py`.

````python
%%writefile click_counter.py
import streamlit as st

if "clicks" not in st.session_state:
    st.session_state.clicks = 0

if st.button("Click me"):
    st.session_state.clicks += 1

st.write(f"Clicks: {st.session_state.clicks}")
````
